# Sample Table Creator

This notebook accepts **catalog** and **schema** as runtime parameters via widgets, creates a sample table, and populates it with random data.

```python
dbutils.widgets.text("catalog", "dbx_dev", "Catalog")
dbutils.widgets.text("schema", "dev_schema", "Schema")

catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")

print(f"Catalog: {catalog}")
print(f"Schema:  {schema}")
```

```python
import random
from datetime import datetime, timedelta

table_name = f"{catalog}.{schema}.sample_data"

# Create the table only if it does not already exist
if spark.catalog.tableExists(table_name):
    print(f"Table {table_name} already exists — skipping creation.")
else:
    # Generate 100 rows of random data
    rows = []
    for i in range(1, 101):
        rows.append((
            i,
            f"item_{i}",
            round(random.uniform(10.0, 1000.0), 2),
            datetime.now() - timedelta(days=random.randint(0, 365)),
        ))

    df = spark.createDataFrame(rows, ["id", "name", "value", "created_at"])
    df.write.mode("overwrite").saveAsTable(table_name)

    print(f"Created table {table_name} with {df.count()} rows.")
```
